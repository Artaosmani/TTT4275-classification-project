```python
"""people = ['m', 'w', 'b', 'g'] # type of peoble, men, women, boy, girl
classes = ['ae', 'ah', 'aw', 'eh', 'ei', 'er', 'ih', 'iy','oa', 'oo', 'uh', 'uw'] # all the different vowels

num_people = 4
num_classes = 12"""

#usikker på oppdeling av trening/testset. Tar bare første 70 som det står i oppgaven foreløpig, men blir da bare menn og kvinner i trening

#
features_cols = [10, 11, 12]  # 11., 12. og 13. kolonne (0-indeksert) Altså de som er 50%? usikker på om riktig

# Dictionary for å lagre data per vokalklasse
class_data = defaultdict(list)

def get_vowel_class(name):
    return name[3:5]  # Henter ut vokalen fra strengen, f.eks. 'ae' fra 'm01ae'

# Les fil + sorter dataene
with open("Data/Iris_TTT4275/vowdata_nohead.dat", "r") as file:
    data = csv.reader(file, delimiter=' ')
    for line in data:
        line = [x for x in line if x]         # Fjern eventuelle tomme elementer (pga. ekstra mellomrom)
        vowel_class = get_vowel_class(line[0])  # Hent vokalklassen
        features = [float(line[i]) for i in features_cols] # Hent de relevante feature-kolonnene og konverter til float
        class_data[vowel_class].append(features)    # Legg til i riktig vokalklasse

print(class_data)
# Konverter til dictionary med lister
class_data = dict(class_data)
print('\n')
print(class_data)
"""
# Skriv ut antall samples per klasse
for vowel, samples in class_data.items():
    print(f"Klasse '{vowel}' har {len(samples)} samples")
    print(samples)
#ser riktig ut egt
"""
```

```python
# Lagring av trenings- og testdata
train_data = {}
test_data = {}

# Del opp dataene
for vowel, data in class_data.items():
    data_array = np.array(data)  # Konverter til NumPy-array
    train_data[vowel] = data_array[:70]  # Første 70 til trening
    test_data[vowel] = data_array[70:]   # Siste 69 til test

# Beregn middelverdi og kovarians for treningssettet
mean_vectors = {}
cov_matrices = {}

for vowel, data in train_data.items():
    mean_vectors[vowel] = np.mean(data, axis=0)  # Middelverdi (gjennomsnitt per feature)
    cov_matrices[vowel] = np.cov(data, rowvar=False)  # Kovariansmatrise

# Print resultatene
for vowel in mean_vectors:
    print(f"\nVokal: {vowel}")
    print(f"Middelverdi: {mean_vectors[vowel]}")
    print(f"Kovariansmatrise:\n{cov_matrices[vowel]}")

```

```python
import numpy as np
import scipy.stats

# Funksjon for å klassifisere en testprøve med full kovariansmodell
def classify_sample(x, mean_vectors, cov_matrices):
    max_prob = -np.inf #vil finne største sannsynlighet så starter så lavt som mulig og oppdaterer hver gang en sansynlighet er høyere. Kan denne like gjærne være 0?
    best_class = None # vi vil finne denne, setter til none enn så lenge
    
    for vowel, mean in mean_vectors.items():
        cov = cov_matrices[vowel]
        prob = scipy.stats.multivariate_normal.logpdf(x, mean=mean, cov=cov) 
        if prob > max_prob:
                max_prob = prob
                best_class = vowel
    return best_class

# Bygg forvirringsmatrisen
confusion_matrix = {v: {v2: 0 for v2 in mean_vectors} for v in mean_vectors}

correct = 0
total = 0

for vowel, test_samples in test_data.items():
    for sample in test_samples:
        predicted = classify_sample(sample, mean_vectors, cov_matrices)
        confusion_matrix[vowel][predicted] += 1
        if predicted == vowel:
            correct += 1
        total += 1

# Print forvirringsmatrisen
print("\nConfusion Matrix:")
for v, row in confusion_matrix.items():
    print(f"{v}: {list(row.values())}")

# Beregn feilrate
error_rate = 1 - (correct / total)
print(f"\nFeilrate: {error_rate:.2%}")
```
