Fitting of model figure for repport

```python
import numpy as np #
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
```

```python
# Generate example data
np.random.seed(0)
X = np.sort(np.random.rand(20))  # 20 points between 0 and 1
y = np.sin(2 * np.pi * X) + 0.1 * np.random.randn(20)  # True function + noise

# Define input range for plotting
X_plot = np.linspace(0, 1, 100).reshape(-1, 1)

# Model configurations: (degree, label)
models = [
    (1, "Underfitting"),
    (4, "Good Fit"),
    (15, "Overfitting")
]

# Plot
plt.figure(figsize=(15, 4))
for i, (degree, title) in enumerate(models):
    plt.subplot(1, 3, i + 1)
    model = make_pipeline(PolynomialFeatures(degree), LinearRegression())
    model.fit(X.reshape(-1, 1), y)
    y_plot = model.predict(X_plot)
    
    plt.scatter(X, y, color='black', label='Data')
    plt.plot(X_plot, y_plot, color='hotpink', linewidth=2)
    plt.title(title)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.ylim(-1.5, 1.5)
    plt.legend()

plt.tight_layout()
plt.show()
```
